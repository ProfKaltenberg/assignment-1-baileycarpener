```python
def main(): 
    api_service_name = "youtube"
    api_version = "v3"
    DEVELOPER_KEY = "AIzaSyAeOMGKFtTTVjxz5opGfJxoFdDQFMk_bsY"

    youtube = googleapiclient.discovery.build(
        api_service_name, api_version, developerKey = DEVELOPER_KEY)

    request = youtube.search().list(
        part="snippet",
        channelType="any",
        eventType="completed",
        maxResults=50,
        order="date",
        q="omad diet", #changed search term to omad diet, referring to 'one meal a day' diets
        relevanceLanguage="en",
        safeSearch="none",
        type="video"
    ) #also removed the topic parameter to broaden returned results
    response = request.execute()
    
    video_data2 = [] #creates list under new name
    
    for item in response['items']:
        video_info2 = {
            'video_id': item['id']['videoId'],
            'title': item['snippet']['title'],
            'channel_title': item['snippet']['channelTitle'],
            'published_at': item['snippet']['publishedAt'],
            'description': item['snippet']['description']
        }
        video_data2.append(video_info2)
    
    # Create a pandas DataFrame
    df2 = pd.DataFrame(video_data2)
    
    return df2
```

```python
df2['published_at'] = pd.to_datetime(df2['published_at'], errors='raise', utc=True) #converts variable 'published_at' into a date/time variable
df2['published_at'] = df2['published_at'].dt.strftime('%Y-%m') #formats the variable into year - month

df2 #display dataframe to check work
```

```python
freddata = pd.read_csv('/Users/baileycarpenter/Downloads/CORESTICKM159SFRBATL (1).csv') #import data from fred
freddata['observation_date'] = pd.to_datetime(freddata['observation_date']).dt.strftime('%Y-%m')#convert observation_date to just year and month format
```

```python
# Merge df2 and freddata dataframes using different column names for dates
# Use left_on and right_on when column names differ between dataframes
merged_df = pd.merge(
    df2, 
    freddata, 
    left_on='published_at',  # Column from df2
    right_on='observation_date',  # Column from freddata
    how='inner'
)
```
